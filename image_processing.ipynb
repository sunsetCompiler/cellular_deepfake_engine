Converting files into .PNG

```python
# Opening Folders
folders = ['BBC020_v1_outlines_cells']
output = 'Test'
os.makedirs(output, exist_ok=True) # Creating output folder

for folder in folders:
    for file in os.listdir(folder):
        if file.lower().endswith('.tif'):
            img = Image.open(os.path.join(folder, file))
            out_name = file.rsplit('.', 1)[0] + '.png'
            img.save(os.path.join(output, out_name))
```

Changing the dimensions of the Images

```python
# Opening Folder
folder = 'Test'
sizes = [] # [(w1, h1), (w2, h2)]       E.g. sizes = [(640, 480), (800, 600), (1024, 768), ...]

# Collecting all dimensions 
for file in os.listdir(folder):
    if file.lower().endswith('.png'):
        path = os.path.join(folder, file)
        with Image.open(path) as img:
            sizes.append(img.size)

#Calculating average dimensions
total_width = sum(w for w, h in sizes)
total_height = sum(h for w, h in sizes)
count = len(sizes)

avg_width = int(total_width / count)
avg_height = int(total_height / count)
target_size = (avg_width, avg_height)


# Resizing
for file in os.listdir(folder):
    if file.lower().endswith('.png'):
        path = os.path.join(folder, file)
        with Image.open(path) as img:
            resized = img.resize(target_size)
            resized.save(path)
```
